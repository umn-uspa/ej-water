```python
import numpy as np
import pandas as pd
import geopandas as gpd
import matplotlib.pyplot as plt
import matplotlib.ticker as ticker
import sys
import os

sys.path.append(os.path.abspath(".."))
from utils.plot_style import apply_plot_style
from utils.config import root_dir

apply_plot_style()
```

```python
# configurations
study_periods = ["2008_2012", "2013_2017", "2018_2022"]
# load the data
df = pd.read_csv(
    root_dir + "results/huc12_rsei_toxconc_weighted.csv", dtype={"huc12": str}, index_col=0
)
```

```python
cols = ["2008_2012_TOXCONC", "2013_2017_TOXCONC", "2018_2022_TOXCONC"]
for col in cols:
    df[col] = df[col].fillna(0)
    df[f"{col}_log"] = np.log(df[col] + 1)
cols = ["2008_2012_TOXCONC_log", "2013_2017_TOXCONC_log", "2018_2022_TOXCONC_log"]
regions = [
    "Headwaters",
    "Gorge",
    "Driftless",
    "Working River",
    "Confluence",
    "Chickasaw",
    "Delta",
    "Lower Mississippi",
    "Gulf South",
]
# layout: 9 rows (regions) and 3 columns (periods)
fig, axes = plt.subplots(
    9, 3, figsize=(7, 5), sharex=True, sharey=True, layout="constrained"
)
for i, region in enumerate(regions):
    for j, col_name in enumerate(cols):
        region_data = df[df["Region"] == region]
        ax = axes[i, j]
        ax.hist(
            region_data[col_name],
            color="grey",
            edgecolor="white",
            lw=0.1,
            log=True,
            bins=20,
            range=(0, 10),
        )

        if i == 0:
            ax.set_title(col_name.split("_TOXCONC")[0].replace("_", " - "))
        if j == 2:
            ax.set_ylabel(region, rotation=0, labelpad=5, ha="left")
            ax.yaxis.set_label_position("right")

        ax.yaxis.set_major_formatter(
            ticker.FuncFormatter(lambda x, p: format(int(x), ","))
        )
        ax.xaxis.set_major_formatter(
            ticker.FuncFormatter(lambda x, p: format(int(x), ","))
        )
        ax.yaxis.set_tick_params(which="minor", left=False)

fig.supxlabel("log of HUC-12 toxicity values")
plt.show()

fig.savefig(f"{root_dir}/figures/Figure S2.pdf")
```
